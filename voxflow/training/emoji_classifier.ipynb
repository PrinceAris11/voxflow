# VoxFlow emoji classifier
This notebook trains a lightweight emoji classifier from text and emotion-like cues. It uses sentence embeddings and a small classifier so it can be fine-tuned quickly on Kaggle.


```python
# Install dependencies
# %pip install -q sentence-transformers scikit-learn pandas

import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report

EMOJI_LABELS = ["😀", "🙂", "😢", "😠", "😎", "❤️"]

rows = [
    ("I am very happy today", "😀"),
    ("This is a pleasant update", "🙂"),
    ("I feel sad and tired", "😢"),
    ("This is really frustrating", "😠"),
    ("That looks amazing and polished", "😎"),
    ("I love this idea", "❤️"),
]

frame = pd.DataFrame(rows, columns=["text", "emoji"])
model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = model.encode(frame["text"].tolist())

clf = LogisticRegression(max_iter=4000, random_state=42)
clf.fit(embeddings, frame["emoji"])

sample_texts = ["I am excited about the launch", "This made me feel down", "This is brilliant and cool"]
encoded = model.encode(sample_texts)
preds = clf.predict(encoded)
print(preds)
```
